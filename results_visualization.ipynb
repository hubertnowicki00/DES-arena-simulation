```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

file_path = ' '
df = pd.read_excel(file_path, sheet_name='AcrossReplicationsSummary')


tally_df = df.iloc[1:, 0:13].copy()
tally_df.columns = df.iloc[0, 0:13]
tally_df = tally_df.dropna(subset=['Name'])
tally_df = tally_df[tally_df['Name'] != '(blank)']

tally_plot_df = tally_df[tally_df['Name'].str.contains('Queue')].copy()
tally_plot_df['Average Of Replication Averages'] = pd.to_numeric(tally_plot_df['Average Of Replication Averages'])
tally_plot_df['Half-Width'] = pd.to_numeric(tally_plot_df['Half-Width'])
tally_plot_df = tally_plot_df.sort_values('Average Of Replication Averages', ascending=False)

sns.set(style="white")
plt.figure(figsize=(10, 6))
plt.bar(tally_plot_df['Name'], tally_plot_df['Average Of Replication Averages'], 
        yerr=tally_plot_df['Half-Width'], capsize=5, color='skyblue', edgecolor='black')
plt.xticks(rotation=45, ha='right')
plt.title('Average Waiting Time per Queue (with 95% Confidence Intervals)', fontsize=14)
plt.ylabel('Average Waiting Time', fontsize=12)
plt.xlabel('Queue Name', fontsize=12)
plt.tight_layout()
plt.savefig('AVG_wt.png', dpi=300)
plt.close()

```

```python
timep_df = df.iloc[1:, 14:25].copy()
timep_df.columns = df.iloc[0, 14:25]
timep_df['Name'] = timep_df['Name'].ffill() 
timep_df = timep_df.dropna(subset=['Type', 'Average Of Replication Averages'])

timep_util_df = timep_df[timep_df['Type'] == 'Instantaneous Utilization'].copy()
timep_util_df = timep_util_df[timep_util_df['Average Of Replication Averages'] > 0]
timep_util_df['Average Of Replication Averages'] = pd.to_numeric(timep_util_df['Average Of Replication Averages'])
timep_util_df['Half-Width'] = pd.to_numeric(timep_util_df['Half-Width'])
timep_util_df = timep_util_df.sort_values('Average Of Replication Averages', ascending=False)

plt.figure(figsize=(10, 6))
plt.bar(timep_util_df['Name'], timep_util_df['Average Of Replication Averages'], 
        yerr=timep_util_df['Half-Width'], capsize=5, color='lightcoral', edgecolor='black')
plt.xticks(rotation=45, ha='right')
plt.title('Average Resource Utilization (with 95% Confidence Intervals)', fontsize=14)
plt.ylabel('Average Instantaneous Utilization', fontsize=12)
plt.xlabel('Resource Name', fontsize=12)
plt.tight_layout()
plt.savefig('AVG_util.png', dpi=300)
plt.close()
```
